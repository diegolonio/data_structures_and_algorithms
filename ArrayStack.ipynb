```python
from array_stack import ArrayStack
```

```python
string = "".join(input().strip().replace(" ", "").split())

delimiters = {"}": "{", "]": "[", ")": "("}
stack = ArrayStack()
is_balanced = True

for index, char in enumerate(string):
    if char in delimiters.values():
        stack.push((char, index))
    elif char in delimiters.keys():
        if stack.is_empty():
            is_balanced = False
            break

        if stack.pop()[0] != delimiters[char]:
            is_balanced = False
            break

if stack.is_empty() and is_balanced:
    print("balanced")
else:
    print("not balanced")
```
