```python
from typing import Any

class Empty(Exception):
    pass

class ArrayStack:
    def __init__(self, capacity: int = 10) -> None:
        self._data: list[Any] = [None] * capacity
        self._capacity: int = capacity
        self._top: int = -1

    def __str__(self) -> str:
        if self.is_empty():
            return "Stack is empty"

        return f"{self._data[:self._top + 1]}"

    def __len__(self) -> int:
        return self._top + 1

    def is_empty(self) -> bool:
        return self._top == -1

    def is_full(self):
        return self._top + 1 == self._capacity

    def top(self):
        if self.is_empty():
            raise Empty("Stack is empty")

        return self._data[self._top]

    def push(self, item: Any) -> None:
        if self.is_full():
            new_stack = [None] * self._capacity * 2

            for index in range(self._top + 1):
                new_stack[index] = self._data[index]

            self._data = new_stack
            self._capacity *= 2

        self._top += 1
        self._data[self._top] = item

    def pop(self):
        if self.is_empty():
            raise Empty("Stack is empty")

        item = self._data[self._top]
        self._data[self._top] = None
        self._top -= 1

        if self._capacity > 10 and (self._top + 1) <= (self._capacity // 4):
            new_stack = [None] * (self._capacity // 2)

            for index in range(self._top + 1):
                new_stack[index] = self._data[index]

            self._data = new_stack
            self._capacity //= 2

        return item

class ArrayStackPythonic:
    def __init__(self) -> None:
        self._data: list[Any] = []

    def __len__(self) -> int:
        return len(self._data)

    def __str__(self) -> str:
        if self.is_empty():
            return "Stack is empty"

        return str(self._data)

    def is_empty(self) -> bool:
        return len(self._data) == 0

    def push(self, item: Any) -> None:
        self._data.append(item)

    def top(self) -> Any:
        if self.is_empty():
            raise Empty("Stack is empty.")

        return self._data[-1]

    def pop(self) -> Any:
        if self.is_empty():
            raise Empty("Stack is empty.")

        return self._data.pop()
```

```python
string = "".join(input().strip().replace(" ", "").split())

delimiters = {"}": "{", "]": "[", ")": "("}
stack = ArrayStack()
is_balanced = True

for index, char in enumerate(string):
    if char in delimiters.values():
        stack.push((char, index))
    elif char in delimiters.keys():
        if stack.is_empty():
            is_balanced = False
            break

        if stack.pop()[0] != delimiters[char]:
            is_balanced = False
            break

if stack.is_empty() and is_balanced:
    print("balanced")
else:
    print("not balanced")
```
