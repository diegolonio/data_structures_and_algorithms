### Using array-based sequences

```python
from data_structures.dynamic_array import DynamicArray
```

```python
class GameEntry:
    """Represents one entry of a list of high scores."""

    def __init__(self, name: str, score: int) -> None:
        self._name = name
        self._score = score

    @property
    def name(self) -> str:
        return self._name

    @property
    def score(self) -> int:
        return self._score

    def __str__(self) -> str:
        return f"({self._name}, {self._score})"


class ScoreBoard:
    """Bounded sequence of high scores in nonincreasing order."""

    def __init__(self, capacity: int = 10) -> None:
        """Initialize scoreboard with given maximum capacity."""
        self._board = DynamicArray[GameEntry](capacity=capacity)
        self._capacity = capacity

    def __getitem__(self, k: int) -> GameEntry:
        return self._board[k]

    def __str__(self) -> str: 
        """Return string representation of the high score list."""
        return "\n".join(str(self._board[j]) for j in range(len(self._board)))

    def add(self, entry: GameEntry) -> None:
        """Consider adding entry to high scores."""
        score = entry.score

        # Does the new entry qualify as a high score?
        good = len(self._board) < self._capacity or score > self._board[-1].score

        if good:
            if len(self._board) < self._capacity:
                self._board.append(entry)
            else:
                self._board[-1] = entry

            j = len(self._board) - 1 # Start at the position of the new entry
            while j > 0 and self._board[j - 1].score < score:
                self._board[j], self._board[j - 1] = self._board[j - 1], self._board[j]
                j -= 1


board = ScoreBoard()
board.add(GameEntry("Jack", 510))
board.add(GameEntry("Mike", 1105))
board.add(GameEntry("Paul", 720))
board.add(GameEntry("Anna", 660))
board.add(GameEntry("Rob", 750))
board.add(GameEntry("Rose", 590))
board.add(GameEntry("Jill Valentine", 740))
board.add(GameEntry("Claudette Morel", 2430))
board.add(GameEntry("Feng Min", 340))
board.add(GameEntry("Rebecca Chávez", 600))
print(board)
```

### Using stack-based arrays

```python
from data_structures.array_stack import ArrayStack
```

```python
string = "".join(input().split())

delimiters = {"}": "{", "]": "[", ")": "("}
stack = ArrayStack[str]()
is_balanced = True

for char in string:
    if char in delimiters.values():
        stack.push(char)
    elif char in delimiters.keys():
        if stack.is_empty():
            is_balanced = False
            break

        if stack.pop() != delimiters[char]:
            is_balanced = False
            break

if stack.is_empty() and is_balanced:
    print("balanced")
else:
    print("not balanced")
```
